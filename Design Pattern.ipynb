# Singleton Pattern:
The singleton pattern is a design pattern that ensures a class has only one instance and provides a global point of access to that instance. This means that regardless of how many times the class is instantiated, there will only ever be one instance of it throughout the application

# Implementation:

```python
class Singleton:
    _instance = None

    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)

        return cls._instance


obj1 = Singleton()
obj2 = Singleton()

print(obj1)
print(obj2)

print(obj1 is obj2)
```

# Thread Safe implementation

While the above implementation of the singleton pattern is straightforward, it is not thread-safe. In a multithreaded environment, multiple threads could simultaneously check if the instance exists, leading to the creation of multiple instances.

To make the singleton pattern thread-safe, we can use synchronization mechanisms such as locks to ensure that only one thread can create the instance at a time.

```python
import threading


class Singleton:
    _instance = None
    _lock = threading.Lock()

    def __new__(cls):
        with cls._lock:
            if cls._instance is None:
                cls._instance = super().__new__(cls)

        return cls._instance


obj1 = Singleton()
obj2 = Singleton()

print(obj1 is obj2)
```

# Advantages: 

1. Single Instance: Ensures that a class has only one instance, which can be useful for managing global resources or maintaining a unique state.

2. Global Access: Provides a global point of access to the instance, allowing components to easily interact with it without needing to pass references around.

3. Lazy Initialization: One of the key advantages of the Singleton Pattern is lazy initialization, where the singleton object is initialized only when it’s requested for the first time. This is helpful when the creation of an object is expensive.

### Disadvantages

1. Violation of Object-Oriented Principles: One of the core principles of object-oriented programming (OOP) is the use of objects to encapsulate state and behavior. A singleton, by enforcing a single instance, can violate the principle of modularity and reuse.
   
2. Inheritance Issues: Singletons pose a problem with inheritance. If a class is designed as a singleton, it restricts the ability to inherit from it and create multiple instances. This limitation goes against the flexibility that OOP provides through inheritance and polymorphism.

3. Testing Challenges: Testing code becomes challenging when we’re unable to create new instances for each test.

4. Multithreading Issues: In a multithreaded environment, ensuring that a singleton remains a single instance while being accessed from multiple threads can lead to complex synchronization issues.


# <center> Factory Pattern </center>

# Factory Design Pattern : 

The Factory Pattern is a creational design pattern used when you want to create objects without exposing the object-creation logic to the code that uses those objects.

it defines an interface for creating objects but lets subclasses decide which object to instantiate.

## Without Factory Pattern : 

```python
class Developer:
    def work(self):
        print("Writing code")


class Tester:
    def work(self):
        print("Testing application")


employee_type = "developer"

if employee_type == "developer":
    employee = Developer()
elif employee_type == "tester":
    employee = Tester()

employee.work()
```

## Problem: 

The problem is that the code responsible for using the employee also needs to know how to create the employee.

As the number of employee types grows: below logic becomes harder to maintain.


## The Factory Pattern : 

```python
# We can move the creation logic into a factory.
    
class Developer:
    def work(self):
        print("Writing code")


class Tester:
    def work(self):
        print("Testing application")


class EmployeeFactory:

    @staticmethod
    def create_employee(employee_type):
        if employee_type == "developer":
            return Developer()
        elif employee_type == "tester":
            return Tester()
        else:
            raise ValueError("Invalid employee type")

# Now the client code becomes:

employee = EmployeeFactory.create_employee("developer")

employee.work()
```

### Note : 

The client doesn't need to know how the object needs to be created:

Developer() </br>
Tester()

It simply asks the factory:

EmployeeFactory.create_employee(...)
